```python

# VERSION 1 


print("VERSION 1 : CALCUL DES MOYENNES DES UE")


#Calcul pour l'UE 1
print("--- UE 1 ---")
nb_notes_ue1 = int(input("Nombre de matières pour l'UE 1 : "))
somme_notes_ue1 = 0
somme_coef_ue1 = 0

#Calcul pour l'UE 1
for i in range(nb_notes_ue1):
    note = float(input(f"Entrez la note {i+1} : "))
    coef = float(input(f"Entrez le coefficient de cette note : "))
    somme_notes_ue1 += note * coef
    somme_coef_ue1 += coef

ue1 = somme_notes_ue1 / somme_coef_ue1 if somme_coef_ue1 > 0 else 0

#Calcul pour l'UE 2
print("--- UE 2 ---")
nb_notes_ue2 = int(input("Nombre de matières pour l'UE 2 : "))
somme_notes_ue2 = 0
somme_coef_ue2 = 0

for i in range(nb_notes_ue2):
    note = float(input(f"Entrez la note {i+1} : "))
    coef = float(input(f"Entrez le coefficient de cette note : "))
    somme_notes_ue2 += note * coef
    somme_coef_ue2 += coef

ue2 = somme_notes_ue2 / somme_coef_ue2 if somme_coef_ue2 > 0 else 0

#Calcul pour l'UE 3
print("--- UE 3 ---")
nb_notes_ue3 = int(input("Nombre de matières pour l'UE 3 : "))
somme_notes_ue3 = 0
somme_coef_ue3 = 0

for i in range(nb_notes_ue3):
    note = float(input(f"Entrez la note {i+1} : "))
    coef = float(input(f"Entrez le coefficient de cette note : "))
    somme_notes_ue3 += note * coef
    somme_coef_ue3 += coef

ue3 = somme_notes_ue3 / somme_coef_ue3 if somme_coef_ue3 > 0 else 0

# Affichage des moyennes finales
print("--- RÉSULTATS VERSION 1 ---")
print(f"Moyenne UE 1 : {round(ue1, 2)}")
print(f"Moyenne UE 2 : {round(ue2, 2)}")
print(f"Moyenne UE 3 : {round(ue3, 2)}")
```

```python
import numpy as np


# VERSION 2 


print("VERSION 2 : CALCUL AVEC LISTES ET COEFFICIENTS ENCADRÉS")


matiere = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]

#Définition des coefficients pour chaque UE (exactement 21 valeurs chacun)
#creation list of "coef*"
coef1 = [10, 10, 7, 7, 0, 5, 0, 6, 0, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]
coef2 = [4, 0, 2, 8, 6, 0, 0, 0, 0, 5, 5, 2, 9, 9, 3, 0, 0, 29, 0, 0, 7]
coef3 = [4, 0, 2, 0, 0, 5, 15, 6, 4, 5, 5, 2, 0, 0, 3, 0, 0, 0, 20, 20, 7]


notes = []
for elt in matiere:
    val = float(input(f"Quelle est votre note pour {elt} ? "))
    notes.append(val)


UE1 = np.average(notes, weights=coef1)
UE2 = np.average(notes, weights=coef2)
UE3 = np.average(notes, weights=coef3)

#Affichage des résultats
#show the resulte rounded
print("\n--- RÉSULTATS VERSION 2 ---")
print(f"Moyenne UE1 : {round(UE1, 2)}")
print(f"Moyenne UE2 : {round(UE2, 2)}")
print(f"Moyenne UE3 : {round(UE3, 2)}")

```

```python
# VERSION 3 


print("VERSION 3 : CALCUL AVEC DICTIONNAIRES")

#Liste des matières du Semestre 1 (R101 à R115 et SAE11 à SAE16)
#List of Semester 1 matter, create list of subject with "R" and "SAE" in front of matter
matieres = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]

#Dictionnaires des coefficients pour chaque UE
#Manual creation of the dictionary
dico_coef_ue1 = {
    "R101": 10, "R102": 10, "R103": 7, "R104": 7, "R105": 0, "R106": 5, "R107": 0,
    "R108": 6, "R109": 0, "R110": 5, "R111": 4, "R112": 2, "R113": 5, "R114": 5,
    "R115": 0, "SAE11": 20, "SAE12": 20, "SAE13": 0, "SAE14": 0, "SAE15": 0, "SAE16": 7
}

dico_coef_ue2 = {
    "R101": 4, "R102": 0, "R103": 2, "R104": 8, "R105": 6, "R106": 0, "R107": 0,
    "R108": 0, "R109": 0, "R110": 5, "R111": 5, "R112": 2, "R113": 9, "R114": 9,
    "R115": 3, "SAE11": 0, "SAE12": 0, "SAE13": 29, "SAE14": 0, "SAE15": 0, "SAE16": 7
}

dico_coef_ue3 = {
    "R101": 4, "R102": 0, "R103": 2, "R104": 0, "R105": 0, "R106": 5, "R107": 15,
    "R108": 6, "R109": 4, "R110": 5, "R111": 5, "R112": 2, "R113": 0, "R114": 0,
    "R115": 3, "SAE11": 0, "SAE12": 0, "SAE13": 0, "SAE14": 20, "SAE15": 20, "SAE16": 7
}

#Dictionnaire pour stocker les notes saisies par l'utilisateur
#make a empty dictionary and request information of "note" to user and add these value in the empty dictionary
dico_notes = {}
print("Veuillez saisir les notes pour chaque ressource/SAÉ :")
for m in matieres:
    note = float(input(f"Note pour {m} : "))
    dico_notes[m] = note

#Fonction pour calculer la moyenne d'une UE à partir des dictionnaires
#creation of definition to calculate the average of ue* who came in "coefs". 
#The "for" loop will associate the coefficient values from the dictionary selected in the "coefs" variable, while "notes" will represent the grades entered into "dico_notes".


def calculer_moyenne_ue(notes, coefs):
    somme_notes = 0.0
    somme_coefs = 0.0

    for matiere, coef in coefs.items():
        if coef > 0:
            somme_notes += notes[matiere] * coef
            somme_coefs += coef

    return somme_notes / somme_coefs if somme_coefs > 0 else 0

#Calcul des moyennes pour chaque UE
#calculate the average with the previous made definition 
moy_ue1 = calculer_moyenne_ue(dico_notes, dico_coef_ue1)
moy_ue2 = calculer_moyenne_ue(dico_notes, dico_coef_ue2)
moy_ue3 = calculer_moyenne_ue(dico_notes, dico_coef_ue3)

#Affichage des résultats
#show the resulte rounded
print("--- RÉSULTATS VERSION 3 ---")
print(f"Moyenne UE1 (Réseaux)       : {round(moy_ue1, 2)}")
print(f"Moyenne UE2 (Télécoms)      : {round(moy_ue2, 2)}")
print(f"Moyenne UE3 (Programmation) : {round(moy_ue3, 2)}")
```

```python
# VERSION 4 
import matplotlib.pyplot as plt

print("VERSION 4 : GRAPHICS ET DÉCISION DU JURY")

#Liste des matières du Semestre 1 (R101 à R115 et SAE11 à SAE16)
#List of Semester 1 matter, create list of subject with "R" and "SAE" in front of matter
matieres = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]

#Dictionnaires des coefficients pour chaque UE
#Manual creation of the dictionary
dico_coef_ue1 = {
    "R101": 10, "R102": 10, "R103": 7, "R104": 7, "R105": 0, "R106": 5, "R107": 0,
    "R108": 6, "R109": 0, "R110": 5, "R111": 4, "R112": 2, "R113": 5, "R114": 5,
    "R115": 0, "SAE11": 20, "SAE12": 20, "SAE13": 0, "SAE14": 0, "SAE15": 0, "SAE16": 7
}

dico_coef_ue2 = {
    "R101": 4, "R102": 0, "R103": 2, "R104": 8, "R105": 6, "R106": 0, "R107": 0,
    "R108": 0, "R109": 0, "R110": 5, "R111": 5, "R112": 2, "R113": 9, "R114": 9,
    "R115": 3, "SAE11": 0, "SAE12": 0, "SAE13": 29, "SAE14": 0, "SAE15": 0, "SAE16": 7
}

dico_coef_ue3 = {
    "R101": 4, "R102": 0, "R103": 2, "R104": 0, "R105": 0, "R106": 5, "R107": 15,
    "R108": 6, "R109": 4, "R110": 5, "R111": 5, "R112": 2, "R113": 0, "R114": 0,
    "R115": 3, "SAE11": 0, "SAE12": 0, "SAE13": 0, "SAE14": 20, "SAE15": 20, "SAE16": 7
}

#Dictionnaire pour stocker les notes saisies par l'utilisateur
#make a empty dictionary and request information of "note" to user and add these value in the empty dictionary
dico_notes = {}
print("Veuillez saisir les notes pour chaque ressource/SAÉ :")
for m in matieres:
    note = float(input(f"Note pour {m} : "))
    dico_notes[m] = note

#Fonction pour calculer la moyenne d'une UE à partir des dictionnaires
#creation of definition to calculate the average of ue* who came in "coefs". 
#The "for" loop will associate the coefficient values from the dictionary selected in the "coefs" variable, while "notes" will represent the grades entered into "dico_notes".


def calculer_moyenne_ue(notes, coefs):
    somme_notes = 0.0
    somme_coefs = 0.0

    for matiere, coef in coefs.items():
        if coef > 0:
            somme_notes += notes[matiere] * coef
            somme_coefs += coef

    return somme_notes / somme_coefs if somme_coefs > 0 else 0

#Calcul des moyennes pour chaque UE
#calculate the average with the previous made definition 
moy_ue1 = calculer_moyenne_ue(dico_notes, dico_coef_ue1)
moy_ue2 = calculer_moyenne_ue(dico_notes, dico_coef_ue2)
moy_ue3 = calculer_moyenne_ue(dico_notes, dico_coef_ue3)

moyennes = [moy_ue1, moy_ue2, moy_ue3]


#Détermination de la décision du jury selon le règlement
#Règle : comptage des UE >= 10
ue_validees = sum(1 for m in moyennes if m >= 10)
ue_entre_8_et_10 = sum(1 for m in moyennes if 8 <= m < 10)
ue_sous_8 = sum(1 for m in moyennes if m < 8)

if ue_validees == 3:
    decision = "Année validée ! Passage en BUT 2 autorisé."
elif ue_validees == 2 and ue_entre_8_et_10 == 1:
    decision = "Année non validée, mais passage en BUT 2 autorisé (UE à valider obligatoirement en BUT 2)."
elif ue_validees == 2 and ue_sous_8 == 1:
    decision = "Année non validée et passage en BUT 2 refusé. Redoublement possible sur décision du conseil."
else:
    decision = "Année non validée et passage en BUT 2 exclu."

print("--- RÉSULTATS ET DÉCISION DU JURY ---")
print(f"Moyenne UE 1 : {moy_ue1}")
print(f"Moyenne UE 2 : {moy_ue2}")
print(f"Moyenne UE 3 : {moy_ue3}")
print(f"\nDécision du jury : {decision}")


ues = ['UE 1 (Réseaux)', 'UE 2 (Télécoms)', 'UE 3 (Prog)']
couleurs = ['green' if m >= 10 else 'orange' if m >= 8 else 'red' for m in moyennes]

plt.figure(figsize=(8, 5))
bars = plt.bar(ues, moyennes, color=couleurs, width=0.5)

#Lignes de référence pour le jury 
#Reference lines for the jury
plt.axhline(y=10, color='blue', linestyle='--', label='Seuil de validation (10)')
plt.axhline(y=8, color='red', linestyle=':', label='Seuil éliminatoire (8)')

# Affichage de la valeur sur chaque barre
# Displaying the value on each bar
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 0.3, f"{yval:.2f}", ha='center', va='bottom', fontweight='bold')

plt.ylim(0, 20)
plt.ylabel('Moyenne / 20')
plt.title(f"Moyennes des UE - Semestre 1\nDécision : {decision}", fontsize=10, color='darkblue')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

```python
# VERSION 5

from tkinter import *
from tkinter import ttk
import matplotlib.pyplot as plt 

# Dictionnaire des coefficients du Semestre 1 (UE1, UE2, UE3) pour chaque matière
# Dictionary of Semester 1 coefficients (UE1, UE2, UE3) for each subject
matieres = {
    "R101": [10, 4, 4], "R102": [10, 0, 0], "R103": [7, 2, 2],
    "R104": [7, 8, 0], "R105": [0, 6, 0], "R106": [5, 0, 5],
    "R107": [0, 0, 15], "R108": [6, 0, 6], "R109": [0, 0, 4],
    "R110": [5, 5, 5], "R111": [4, 5, 5], "R112": [2, 2, 2],
    "R113": [5, 9, 0], "R114": [5, 9, 0], "R115": [0, 3, 3],
    "SAE11": [20, 0, 0], "SAE12": [20, 0, 0], "SAE13": [0, 29, 0],
    "SAE14": [0, 0, 20], "SAE15": [0, 0, 20], "SAE16": [7, 7, 7]
}

# Initialisation de la fenêtre principale Tkinter
# Initialization of the main Tkinter window
fenetre = Tk()
fenetre.title("Calculateur d'UE - BUT R&T S1")
fenetre.geometry("620x750")

entrees = {} 

# Génération dynamique des lignes de saisie pour chaque matière
# Dynamic generation of input lines for each subject
ligne = 0
for nom in matieres.keys():
    label = ttk.Label(fenetre, text=nom, font=("Arial", 9, "bold"))
    label.grid(row=ligne, column=0, padx=15, pady=2, sticky="w")
   
    saisie = ttk.Entry(fenetre, width=10)
    saisie.grid(row=ligne, column=1, padx=15, pady=2)
    entrees[nom] = saisie
    ligne += 1

# Création des labels pour afficher les résultats textuels de chaque UE
# Creating labels to display the text results of each UE
label_ue1 = ttk.Label(fenetre, text="UE 1 : --", font=("Arial", 11, "bold"), foreground="darkblue")
label_ue1.grid(row=2, column=3, padx=40, sticky="w")

label_ue2 = ttk.Label(fenetre, text="UE 2 : --", font=("Arial", 11, "bold"), foreground="darkblue")
label_ue2.grid(row=5, column=3, padx=40, sticky="w")

label_ue3 = ttk.Label(fenetre, text="UE 3 : --", font=("Arial", 11, "bold"), foreground="darkblue")
label_ue3.grid(row=8, column=3, padx=40, sticky="w")

# Fonction exécutée au clic sur le bouton pour calculer et afficher le graphique
# Function executed when clicking the button to calculate and display the chart
def calculer_moyennes():
    totaux_points = [0.0, 0.0, 0.0]
    totaux_coeffs = [0.0, 0.0, 0.0]
   
    # Récupération et traitement des notes saisies
    # Retrieval and processing of entered grades
    for nom, saisie in entrees.items():
        valeur = saisie.get().strip().replace(',', '.')
        if valeur != "":
            try:
                note = float(valeur)
                coeffs = matieres[nom]
                for i in range(3):
                    totaux_points[i] += note * coeffs[i]
                    totaux_coeffs[i] += coeffs[i]
            except ValueError:
                pass
               
    moyennes = []

    # Calcul de la moyenne de l'UE 1
    # Calculating the average for UE 1
    if totaux_coeffs[0] > 0:
        moy_1 = round(totaux_points[0] / totaux_coeffs[0], 2)
        label_ue1.config(text=f"UE 1 : {moy_1} / 20")
        moyennes.append(moy_1)
    else:
        moyennes.append(0)
       
    # Calcul de la moyenne de l'UE 2
    # Calculating the average for UE 2
    if totaux_coeffs[1] > 0:
        moy_2 = round(totaux_points[1] / totaux_coeffs[1], 2)
        label_ue2.config(text=f"UE 2 : {moy_2} / 20")
        moyennes.append(moy_2)
    else:
        moyennes.append(0)
       
    # Calcul de la moyenne de l'UE 3
    # Calculating the average for UE 3
    if totaux_coeffs[2] > 0:
        moy_3 = round(totaux_points[2] / totaux_coeffs[2], 2)
        label_ue3.config(text=f"UE 3 : {moy_3} / 20")
        moyennes.append(moy_3)
    else:
        moyennes.append(0)

    # Paramétrage et génération du diagramme Matplotlib
    # Setting up and generating the Matplotlib chart
    ues = ['UE 1 (Réseaux)', 'UE 2 (Télécoms)', 'UE 3 (Prog)']
    couleurs = ['green' if m >= 10 else 'orange' if m >= 8 else 'red' for m in moyennes]

    plt.figure(figsize=(8, 5))
    plt.bar(ues, moyennes, color=couleurs, width=0.5)
    plt.axhline(y=10, color='blue', linestyle='--', label='Seuil de validation (10)')
    plt.axhline(y=8, color='red', linestyle=':', label='Seuil éliminatoire (8)')
    plt.ylim(0, 20)
    plt.ylabel('Moyenne / 20')
    plt.title("Moyennes des UE - Semestre 1", fontsize=11, color='darkblue')
    plt.legend()
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()

# Création du bouton de validation lié à la fonction
# Creating the submit button linked to the function
bouton = ttk.Button(fenetre, text="Calculer et Afficher le Graphique", command=calculer_moyennes)
bouton.grid(row=ligne, column=0, columnspan=2, pady=15)

fenetre.mainloop()
```
